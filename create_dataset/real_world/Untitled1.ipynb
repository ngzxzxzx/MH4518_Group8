```python
stock = 'DIS'
month = '2602'
for i in range(65,155,5):
    for d in range(1,32):
        if d<10:
            date = f'{month}0{d}'
        else:
            date = f'{month}{d}'
        if i >= 100:
            t_str = f'{stock}{date}C00{i}000'
        else:
            t_str = f'{stock}{date}C000{i}000'

        opt = yf.Ticker(t_str)
        df = opt.history('1y')
        print(df.head())
```
